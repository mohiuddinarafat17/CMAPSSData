```python
!pip install shap tensorflow pandas numpy matplotlib seaborn scikit-learn -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, LSTM, Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping
import shap

print("✅ Libraries installed!")
```

```python
columns = ['unit_id', 'cycle'] + ['setting1', 'setting2', 'setting3'] + [f'sensor{i}' for i in range(1,22)]

train_df = pd.read_csv('/content/train_FD002.txt', sep=r'\s+', header=None, names=columns)
test_df  = pd.read_csv('/content/test_FD002.txt',  sep=r'\s+', header=None, names=columns)
rul_df   = pd.read_csv('/content/RUL_FD002.txt',   sep=r'\s+', header=None, names=['RUL'])

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
```

```python
# <<< CHANGE HERE if you want different cap (usually 125 is best) >>>
RUL_CAP = 125

train_df['RUL'] = train_df.groupby('unit_id')['cycle'].transform(lambda x: x.max() - x)
train_df['RUL'] = train_df['RUL'].clip(upper=RUL_CAP)

print("✅ Piecewise RUL applied with cap =", RUL_CAP)
print(train_df[['unit_id', 'cycle', 'RUL']].tail())
```

```python
# <<< CHANGE HERE >>> - Selected important sensors only (reduces noise)
feature_cols = ['sensor2', 'sensor3', 'sensor4', 'sensor7', 'sensor8', 'sensor9',
                'sensor11', 'sensor12', 'sensor13', 'sensor14', 'sensor15',
                'sensor17', 'sensor20', 'sensor21']

scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_df[feature_cols])
test_scaled  = scaler.transform(test_df[feature_cols])

train_scaled_df = pd.DataFrame(train_scaled, columns=feature_cols)
train_scaled_df['unit_id'] = train_df['unit_id'].values
train_scaled_df['RUL'] = train_df['RUL'].values

test_scaled_df = pd.DataFrame(test_scaled, columns=feature_cols)
test_scaled_df['unit_id'] = test_df['unit_id'].values

print(f"✅ Using {len(feature_cols)} important sensors.")
```

```python
# <<< CHANGE HERE >>> - Increase to 60-80 if you have enough RAM
TIME_STEPS = 50

def create_sequences(data, time_steps):
    X, y = [], []
    for unit in data['unit_id'].unique():
        unit_data = data[data['unit_id'] == unit]
        unit_values = unit_data[feature_cols].values
        unit_rul = unit_data['RUL'].values
        for i in range(len(unit_values) - time_steps):
            X.append(unit_values[i:i + time_steps])
            y.append(unit_rul[i + time_steps])
    return np.array(X), np.array(y)

X_train, y_train = create_sequences(train_scaled_df, TIME_STEPS)

print(f"✅ X_train shape: {X_train.shape}")
```

```python
model = Sequential()

model.add(Conv1D(filters=64, kernel_size=3, activation='relu',
                 input_shape=(TIME_STEPS, len(feature_cols))))
model.add(BatchNormalization())
model.add(MaxPooling1D(pool_size=2))
model.add(Dropout(0.2))

model.add(Conv1D(filters=128, kernel_size=3, activation='relu'))
model.add(BatchNormalization())
model.add(Dropout(0.2))

model.add(LSTM(128, return_sequences=True))
model.add(Dropout(0.3))
model.add(LSTM(64))
model.add(Dropout(0.3))

model.add(Dense(32, activation='relu'))
model.add(Dense(1))

model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()

# <<< CHANGE HERE >>> - You can increase epochs to 200 if needed
early_stop = EarlyStopping(monitor='val_loss', patience=25, restore_best_weights=True)

history = model.fit(X_train, y_train,
                    epochs=150,
                    batch_size=32,
                    validation_split=0.2,
                    callbacks=[early_stop],
                    verbose=1)

plt.figure(figsize=(10,5))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
def create_test_sequences(data, time_steps):
    X_test = []
    for unit in data['unit_id'].unique():
        unit_data = data[data['unit_id'] == unit][feature_cols].values
        if len(unit_data) >= time_steps:
            X_test.append(unit_data[-time_steps:])
        else:
            padding = np.zeros((time_steps - len(unit_data), len(feature_cols)))
            seq = np.vstack((padding, unit_data))
            X_test.append(seq)
    return np.array(X_test)

X_test = create_test_sequences(test_scaled_df, TIME_STEPS)
y_pred = model.predict(X_test).flatten()
y_true = rul_df['RUL'].values

rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)
r2 = r2_score(y_true, y_pred)

print(f"\n🎉 NEW RESULTS AFTER FIXES")
print(f"RMSE : {rmse:.4f}")
print(f"MAE  : {mae:.4f}")
print(f"R²   : {r2:.4f}")
```

```python
results_df = pd.DataFrame({
    'Engine_ID': range(1, len(y_true) + 1),
    'Actual_RUL': y_true,
    'Predicted_RUL': np.round(y_pred, 2),
    'Absolute_Error': np.round(np.abs(y_true - y_pred), 2)
})

print("\n📊 RUL Results for First 20 Engines:")
print(results_df.head(20))

results_df.to_csv('RUL_Predictions_Per_Engine.csv', index=False)
print("\n✅ Full results saved to RUL_Predictions_Per_Engine.csv")
```

```python
background = X_train[np.random.choice(len(X_train), 50, replace=False)]
explainer = shap.GradientExplainer(model, background)

test_sample_original = X_test[:30]
shap_values = explainer.shap_values(test_sample_original)

# Reshape shap_values and test_sample for summary_plot
# shap_values from GradientExplainer is typically (num_samples, time_steps, num_features, 1) for single output
# We need (num_samples, num_features)

# First, remove the redundant dimension of size 1 (the last dimension, axis=-1)
shap_values_squeezed = np.squeeze(shap_values, axis=-1) # Shape becomes (num_samples, time_steps, num_features)

# Then, average over the time_steps dimension (axis=1) for both shap_values and test_sample
shap_values_reshaped = shap_values_squeezed.mean(axis=1) # Shape becomes (num_samples, num_features)

# test_sample_original is (num_samples, time_steps, num_features)
test_sample_reshaped = test_sample_original.mean(axis=1) # Shape becomes (num_samples, num_features)

print("✅ SHAP analysis completed!")

shap.summary_plot(shap_values_reshaped, test_sample_reshaped, feature_names=feature_cols, show=True)
```
